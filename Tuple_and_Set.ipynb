# Tuple
## ordered collection of items ()
## One main difference is tuple is immutable
## var_name = list[[stu1:()],[stu2:()]...]

```python
# Create a Tuple
student=(101,"Sanjiv","CSE",8.9)
print(student)

# Tuple using index
print("Student information in first location:",student[0])
print(student[1])

# negative index
print("Example for negative index:",student[-2])
```

```python
subjects=("OS",'Dsv',"DSA","Python","Machine Learning")
print(len(subjects))
```

```python
# slicing
subjects=("DSV","DSA","Statistics","SQL","Machine Learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
# Tuple Immutable
students_info("Chirag",23,"CSE",8.7)
students_info[3]=9.1
print(students_info)
```

```python
# nested tuple
student=(101,"Rahul",(80,78,66))
print(student)
print(student[2])
print(student[2][1])
```

```python
# count function
marks=(75,78,68,75,89,75,92,98)
print("Time of Occurence:",marks.count(75))
```

```python
# find the element present in tuple or not
subjects=("Python",
          "DSV",
          "SQL",
          "Machine Learning",
          "Statistics")
print("Python" in subjects)
print("Java" in subjects)
```

# SET
# unordered collection of items
# set is mutable
# allow the different data types
# it does not support for index
# it gives output with unique values ( NO DUPLICATION )

```python
# Symbol for set { }
# Create a set with similar items
marks={75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
# create a empty set
a={}
a=set()
# check its type
print(type(a))
```

```python
# set does not support index
students={"Amit","Riya","Priya"}
print(students[1])
```

```python
# add an item in the set
students={"Amit","Kiran","Sneha"}
students.add("Priya")
print(students)
students.update(["Ram","Chitra","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Priya")
print(students)

students.discard("Amit")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.remove("Karan")
print(students)
```

```python
students={"Amit","Riya","Priya"}
students.discard("Karan")
print(students)
```

```python
# create a set and clear the list of items
students_info={"Amit",
              "Karan",
              "Priya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99,88,77,66}
del marks
print(marks)
```

```python
# set operations
python_students={"Amit","Neha","Sanjiv","Manish"}
java_students={"Amit","Karan","Dhilip","Manish","Neha"}

# for union operation by using set

result=python_students.union(java_students)
print("Union Operation:",result)

# for intersection operation by using set

result_intersection=python_students.intersection(java_students)
print("Intersection Operation:",result_intersection)

# for difference operation by using set

result_difference=python_students.difference(java_students)
print("For difference:",result_difference)
```

```python
# convert list to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
unique_roll_no=list(unique_stud_id)
print(unique_roll_no)
```
