# For loop


```python
#for loop
numbers=range(5)
for i in numbers:
  print(i)
```

```python
# simple for loop
for i in range(1,10):
  print(i)
```

```python
# for loop step increment
for i in range(1,10,2):
  print(i)
```

```python
# while loop
i=1
while i<=10:
  print(i)
  i=i+1
```

# List

## Definition

### A list is a collection of ordered items stored in square bracket [ ]
it is mutable (elements can be changed_)

```python
Cities=["Delhi","Chennai","Mumbai"]
```

```python
# create one more list
# print it
marks=[92,96,76,58,85]
print(marks)
```

```python
#list of different data types
student=[101,"Rahul Sharma","Computer Science",
         8.99]
print(student)
```

```python
#Access elements by index
student=[102,"Anil","ECE",8.92]
print(student[0])
print(student[-2])
print(student[0])
print(student[2])
print(student[3])
```

```python
#changing an element in the list
cgpa=[7.2,8.4,6.9,7.8]
print(cgpa)
cgpa[2]=9.1
print("After update")
print(cgpa)
```

```python
# add an element in the list using append
students=["Anish",
          "Bhavya",
          "Chirag"
          ]
students.append("Deepika")
print(students)
```

```python
# add an item in the list in a particular location
students=["Nandhini","Harish","Manisha"]
students.insert(1,"Anish")
print(students)
```

```python
# extend two list
section_a=["Anish","Girish",'Jhansi']
print(section_a)
section_b=["Bheem","Chirag"]
section_a.extend(section_b)
print(section_a)
```

```python
# remove an item in the list
subjects=["Python","java","DBMS","DSA"]
removed=subjects.pop()
print(removed)
```

```python
# to find the length of the list
subjects=["Python","java","DBMS","DSA"]
print(len(subjects))
```

```python
# sorting the elements in the list
marks=[78,99,56,64,63,45,97]
marks.sort()
print(marks)
```

```python
# nested list
marks=[
    [91,88,76],
    [78,56,67],
    [85,89,78]
]
print(marks)
print(marks[0])
print(marks[1][2])
print(marks[2][0])
```

# Tuple

## definition


### A tuple is ordered collection of items
### It is immutable
### TUPLE given in the ()


```python
# create a tuple
student = ("Chirag",21,'CSE',8.7)
print(student)
```

```python
# display the items by index
subjects=("Python","DSA","DAA","OS",'SQL','Statistics')
print(subjects[0])
print(subjects[5])
```

```python
subjects=("Python","DSA","DAA","OS",'SQL','Statistics')
subjects[2]='Algebra'
print(subjects)
```

```python
# tuple slicing and printing the length and find the item is present in the tuple
subjects=("Python",
          'Statistics',
          'SQL',"Machine Leaening","Deep Learning")
print(subjects[1:4])
print("Number Of Subjects:",len(subjects))
print("Deep Learning" in subjects)
print("Relational Algebra" in subjects)
```

```python
# nested tuple
student=(101,'Rahul',(85,98,76))
print(student)
print(student[2])
print(student[2][1])
```
